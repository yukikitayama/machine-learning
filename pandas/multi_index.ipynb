# MultiIndex

```python
import pandas as pd
```

```python
bigmac = pd.read_csv("bigmac.csv", parse_dates=["Date"], date_format="%Y-%m-%d")
bigmac.head()
bigmac.dtypes
bigmac.info()
```

```python
bigmac = pd.read_csv("bigmac.csv", parse_dates=["Date"], date_format="%Y-%m-%d")
bigmac.head()

bigmac.set_index("Date")
bigmac.set_index("Country")
bigmac.set_index(keys=["Date", "Country"])
bigmac.set_index(keys=["Country", "Date"]).sort_index()
bigmac.nunique()
bigmac = bigmac.set_index(keys=["Date", "Country"]).sort_index()

bigmac.head()

bigmac = pd.read_csv("bigmac.csv", parse_dates=["Date"], date_format="%Y-%m-%d", index_col=["Date", "Country"]).sort_index()

bigmac.head()

bigmac.index
bigmac.index.names
bigmac.index[0]
```
